```python
# ============================================================
# UTILITAIRES — NORMALISATION DU TEXTE
# À exécuter AVANT resolve_reference_sources()
# ============================================================

import pandas as pd
import numpy as np
import re

def normalize_text(x):
    """Normalise un texte pour comparer les noms de colonnes/questions."""
    if pd.isna(x):
        return ""
    return str(x).strip().lower()

def normalize_child_id(x):
    """Normalise l'identifiant enfant."""
    if pd.isna(x):
        return ""
    return str(x).strip()

def first_existing(df, candidates):
    """Retourne la première colonne existante parmi les candidates."""
    for col in candidates:
        if col in df.columns:
            return col
    return None

print("✅ Fonctions utilitaires chargées")
print("normalize_text :", callable(normalize_text))
print("normalize_child_id :", callable(normalize_child_id))
print("first_existing :", callable(first_existing))
```

```python
# ============================================================
# DÉTECTION ROBUSTE DES 10 ITEMS DE RÉFÉRENCE
# ============================================================

REFERENCE_ITEM_PATTERNS = {
    "A1": ["During a short activity (less than 10 min)", "short activity", "less than 10 min"],
    "A2": ["After a single clear instruction"],
    "L1": ["pause, use vague words", "can't find the right one", "cant find the right one"],
    "L2": ["Does the child mispronounce sounds that should already be mastered for their age",
          "mispronounce sounds", "pronunciation errors", "speech sounds"],
    "R1": ["mix up letters that look alike", "b/d, p/q"],
    "R2": ["read noticeably slower", "struggle to decode"],
    "S1": ["avoid looking at the speaker's face", "look away, down"],
    "S2": ["struggle to take turns, share", "respond to others' actions"],
    "M1": ["handwriting messy", "press too hard", "tire quickly"],
    "M2": ["hold the pencil awkwardly", "lack of control"]
}

ROLE_ALLOWED = {
    "A1": {"teacher", "nurse"},
    "A2": {"teacher", "nurse", "speech therapist"},
    "L1": {"teacher", "speech therapist"},
    "L2": {"speech therapist"},
    "R1": {"teacher"},
    "R2": {"teacher"},
    "S1": {"teacher", "nurse"},
    "S2": {"teacher", "nurse"},
    "M1": {"teacher", "nurse"},
    "M2": {"nurse"}
}

def resolve_reference_sources():
    """
    Retourne TOUTES les occurrences d'un item dans les tables
    professionnelles. Cela évite de perdre, par exemple, A2
    lorsqu'il existe chez Teacher et Speech therapist.
    """
    sources = {item: [] for item in REFERENCE_ITEM_PATTERNS}

    for table_name, df in [("df2_clean", df2_clean), ("df3_clean", df3_clean)]:
        for item, patterns in REFERENCE_ITEM_PATTERNS.items():
            for col in df.columns:
                col_norm = normalize_text(col)
                score = 0

                for p in patterns:
                    pn = normalize_text(p)
                    if pn and pn in col_norm:
                        score = max(score, len(pn))

                if item == "L2" and "mispronounce sounds" in col_norm:
                    score += 10000

                if score > 0:
                    sources[item].append((table_name, col, score))

    # Dédupliquer les colonnes identiques.
    for item in sources:
        seen = set()
        clean = []
        for table_name, col, score in sorted(
            sources[item], key=lambda z: z[2], reverse=True
        ):
            key = (table_name, col)
            if key not in seen:
                seen.add(key)
                clean.append((table_name, col))
        sources[item] = clean

    return sources

reference_sources = resolve_reference_sources()

print("=" * 72)
print("ITEMS DE RÉFÉRENCE — TOUTES LES SOURCES")
print("=" * 72)

for item in ["A1","A2","L1","L2","R1","R2","S1","S2","M1","M2"]:
    if reference_sources[item]:
        print(f"✅ {item}")
        for table_name, col in reference_sources[item]:
            print(f"     └─ {table_name} → {col}")
    else:
        print(f"⚠️ {item}: NON TROUVÉ")

missing_reference_items = [
    item for item in reference_sources if not reference_sources[item]
]

if missing_reference_items:
    print("\n⚠️ Aucun remplissage artificiel :", missing_reference_items)
else:
    print("\n🎉 Les 10 items sont détectés.")
```

```python
# ============================================================
# FONCTIONS DE RÈGLES — TOUTES LES TABLES PROFESSIONNELLES
# ============================================================

def get_table(table_name):
    return df2_clean if table_name == "df2_clean" else df3_clean

def get_child_observations(child_id):
    target = normalize_child_id(child_id)
    frames = []

    for df in [df2_clean, df3_clean]:
        cid = first_existing(df, CHILD_ID_CANDIDATES)
        if cid is None:
            continue

        temp = df.copy()
        temp["_child_norm"] = temp[cid].map(normalize_child_id)
        temp = temp[temp["_child_norm"] == target].copy()

        if not temp.empty:
            frames.append(temp)

    return frames

def get_family_row(child_id):
    if df1_clean.empty:
        return None

    target = normalize_child_id(child_id)
    temp = df1_clean.copy()
    temp["_child_norm"] = temp[FAMILY_CHILD_ID].map(normalize_child_id)
    rows = temp[temp["_child_norm"] == target]

    return None if rows.empty else rows.iloc[-1]

def item_rows(child_id, item):
    """Toutes les lignes de toutes les tables où l'item existe."""
    target = normalize_child_id(child_id)
    parts = []

    for table_name, col in reference_sources.get(item, []):
        df = get_table(table_name)
        cid = first_existing(df, CHILD_ID_CANDIDATES)
        role = first_existing(df, ROLE_CANDIDATES)
        week = first_existing(df, WEEK_CANDIDATES)

        if cid is None or role is None:
            continue

        cols = [role, col]
        if week:
            cols.append(week)

        sub = df[df[cid].map(normalize_child_id) == target][cols].copy()
        if sub.empty:
            continue

        sub["_table"] = table_name
        sub["_score"] = pd.to_numeric(sub[col], errors="coerce")
        sub["_role"] = sub[role].map(role_normalized)
        if week:
            sub["_week"] = sub[week].astype(str).str.strip()
        else:
            sub["_week"] = ""

        parts.append(sub)

    return pd.concat(parts, ignore_index=True) if parts else pd.DataFrame()

def active_item(child_id, item):
    sub = item_rows(child_id, item)

    if sub.empty:
        return False

    sub = sub.dropna(subset=["_score"])

    allowed = ROLE_ALLOWED.get(item)
    if allowed:
        sub = sub[
            sub["_role"].apply(lambda r: any(a in r for a in allowed))
        ]

    high = sub[sub["_score"] >= 2]
    if high.empty:
        return False

    distinct_roles = high["_role"].nunique()

    if item in {"A1", "A2"}:
        return distinct_roles >= 2 and high["_week"].nunique() >= 3

    if item in {"L1", "S1", "S2"}:
        return distinct_roles >= 2

    if item == "L2":
        return high["_role"].str.contains(
            "speech therapist|orthophon", regex=True
        ).any()

    # M1/M2
    return True

def literacy_active(child_id):
    """
    R-ROUTE-04 :
    - R1 : score >=2 persistant, même enseignant, observations
      espacées d'au moins 3 semaines.
    - R2 : score >=2 de façon persistante.
    """
    r1 = item_rows(child_id, "R1")
    r2 = item_rows(child_id, "R2")

    r1_active = False
    if not r1.empty:
        r1 = r1[
            r1["_role"].str.contains("teacher", regex=True)
            & r1["_score"].ge(2)
        ].copy()
        if not r1.empty:
            # Même enseignant + au moins 3 semaines entre observations.
            for teacher in r1["_role"].unique():
                weeks = pd.to_numeric(
                    r1.loc[r1["_role"] == teacher, "_week"],
                    errors="coerce"
                ).dropna().sort_values().unique()
                if len(weeks) >= 2 and (weeks.max() - weeks.min()) >= 3:
                    r1_active = True
                    break

    r2_active = False
    if not r2.empty:
        r2 = r2[
            r2["_role"].str.contains("teacher", regex=True)
            & r2["_score"].ge(2)
        ].copy()
        r2_active = len(r2) >= 2

    return r1_active or r2_active

def has_discrepancy(child_id):
    """
    R-ROUTE-09 sur toutes les observations disponibles.
    Une contradiction n'est retenue que si deux rôles distincts
    donnent >=2 et 0 sur le même item.
    """
    for item in reference_sources:
        sub = item_rows(child_id, item)

        if sub.empty:
            continue

        high_roles = set(sub.loc[sub["_score"] >= 2, "_role"])
        zero_roles = set(sub.loc[sub["_score"] == 0, "_role"])

        if high_roles and zero_roles and high_roles.isdisjoint(zero_roles):
            return True, item

    return False, None

def yes_value(x):
    return normalize_text(x) in {"yes", "oui", "true", "1"}

def recent_ongoing_event(row, event_col, timing_col, status_col):
    if row is None or not event_col:
        return False

    event = yes_value(row.get(event_col, ""))
    timing = normalize_text(row.get(timing_col, "")) if timing_col else ""
    status = normalize_text(row.get(status_col, "")) if status_col else ""

    recent = timing in {
        "this month", "1–2mo", "1-2mo", "1–2 mo", "1-2 mo"
    }

    return event and recent and status == "ongoing"

def language_confounded(row):
    if row is None:
        return False

    # Le formulaire actuel peut ne pas contenir ce champ.
    # Dans ce cas, aucune confusion linguistique n'est inventée.
    candidates = [
        c for c in df1_clean.columns
        if any(k in normalize_text(c) for k in [
            "home language", "language differs", "langue", "linguistic"
        ])
    ]

    for col in candidates:
        v = normalize_text(row.get(col, ""))
        if v in {"yes", "oui", "true", "1"} or "differs" in v or "different" in v or "différ" in v:
            return True

    return False

def get_age(row):
    if row is None:
        return np.nan

    candidates = [
        c for c in df1_clean.columns
        if "age" in normalize_text(c) or "âge" in normalize_text(c)
    ]

    if not candidates:
        return np.nan

    return pd.to_numeric(row.get(candidates[0]), errors="coerce")
```

```python
# ============================================================
# L2 — VALIDATION PROFESSIONNELLE
# ============================================================

def l2_professional_flag(df_child):
    """
    L2 est actif uniquement si :
    - le score L2 >= 2
    - l'observation vient d'un Speech therapist / orthophoniste
    """
    if "L2" not in item_columns:
        return False

    col = item_columns["L2"]

    sub = df_child[[ROLE_COL, col]].copy()
    sub[col] = pd.to_numeric(sub[col], errors="coerce")
    sub = sub.dropna(subset=[col])

    sub["_role_norm"] = sub[ROLE_COL].map(role_normalized)

    high = sub[sub[col] >= 2]

    if high.empty:
        return False

    return high["_role_norm"].str.contains(
        "speech therapist|orthophon",
        regex=True
    ).any()


print("✅ l2_professional_flag chargé")
```

```python
# ============================================================
# CONFIGURATION — identifiant enfant
# ============================================================

CHILD_ID = "Child ID"
TARGET_COLUMN = "Reference_Result"

print("✅ CHILD_ID =", CHILD_ID)
print("✅ TARGET_COLUMN =", TARGET_COLUMN)
```

```python
# ============================================================
# CONDITIONS DE CONTEXTE POUR LES ROUTES ATTENTION / SOCIAL
# ============================================================

def event_not_recent_resolved_or_absent(row, event_col, timing_col, status_col):
    """
    Condition du référentiel pour R-01 / R-05 :
    - événement absent, OU
    - événement résolu ET non récent.
    Un événement récent résolu n'autorise donc pas la route.
    """
    if row is None or not event_col:
        return True

    event = yes_value(row.get(event_col, ""))
    if not event:
        return True

    timing = normalize_text(row.get(timing_col, "")) if timing_col else ""
    status = normalize_text(row.get(status_col, "")) if status_col else ""

    recent = timing in {
        "this month", "1–2mo", "1-2mo", "1–2 mo", "1-2 mo"
    }

    return status == "resolved" and not recent


# ============================================================
# MOTEUR FINAL — REFERENCE_RESULT
# ============================================================

def build_reference_result():
    # Union de TOUS les enfants présents dans les observations professionnelles.
    child_ids = set()

    for df in [df2_clean, df3_clean]:
        cid = first_existing(df, CHILD_ID_CANDIDATES)
        if cid:
            child_ids.update(
                x for x in df[cid].map(normalize_child_id).unique()
                if x not in {"", "nan", "none", "."}
            )

    records = []

    for child_id in sorted(child_ids):
        family_row = get_family_row(child_id)

        # Signaux bruts
        attention_signal = active_item(child_id, "A1") or active_item(child_id, "A2")
        l1_flag = active_item(child_id, "L1")
        l2_flag = l2_professional_flag(child_id)
        language_flag = l1_flag or l2_flag

        literacy_flag = literacy_active(child_id)
        social_signal = active_item(child_id, "S1") and active_item(child_id, "S2")
        motor_flag = active_item(child_id, "M1") or active_item(child_id, "M2")

        recent_family = recent_ongoing_event(
            family_row, FAMILY_EVENT_COL, FAMILY_TIMING_COL, FAMILY_STATUS_COL
        )
        recent_health = recent_ongoing_event(
            family_row, HEALTH_EVENT_COL, HEALTH_TIMING_COL, HEALTH_STATUS_COL
        )
        recent_context = recent_family or recent_health

        # Les routes Attention/Social ont des conditions de contexte
        # supplémentaires dans le référentiel.
        attention_context_clear = (
            event_not_recent_resolved_or_absent(
                family_row, FAMILY_EVENT_COL, FAMILY_TIMING_COL, FAMILY_STATUS_COL
            )
            and
            event_not_recent_resolved_or_absent(
                family_row, HEALTH_EVENT_COL, HEALTH_TIMING_COL, HEALTH_STATUS_COL
            )
        )

        social_context_clear = event_not_recent_resolved_or_absent(
            family_row, FAMILY_EVENT_COL, FAMILY_TIMING_COL, FAMILY_STATUS_COL
        )

        # Flags utilisables pour les routes indépendantes.
        attention_flag = attention_signal and attention_context_clear
        social_flag = social_signal and social_context_clear

        # R-08 regarde les signaux bruts : un contexte récent et ongoing
        # doit bien dégrader un signal présent.
        any_domain_signal = any([
            attention_signal, language_flag, literacy_flag,
            social_signal, motor_flag
        ])

        # --------------------------------------------------------
        # R-ROUTE-08 : contexte récent/ongoing
        # --------------------------------------------------------
        if recent_context and (
            attention_signal or language_flag or social_signal or motor_flag
        ):
            areas = ""
            if family_row is not None and AFFECTED_AREAS_COL:
                areas = str(family_row.get(AFFECTED_AREAS_COL, "")).strip()

            result = "Monitor 4–6 weeks — situational context declared"
            if areas:
                result += f" — affected areas: {areas}"

            records.append({
                CHILD_ID: child_id,
                TARGET_COLUMN: result,
                "Reference_Rule": "R-ROUTE-08",
                "Reference_Item": ""
            })
            continue

        # --------------------------------------------------------
        # R-ROUTE-09 : contradiction
        # --------------------------------------------------------
        discrepancy, discrepancy_item = has_discrepancy(child_id)

        if discrepancy:
            records.append({
                CHILD_ID: child_id,
                TARGET_COLUMN: "Needs clarification — no orientation",
                "Reference_Rule": "R-ROUTE-09",
                "Reference_Item": discrepancy_item
            })
            continue

        # --------------------------------------------------------
        # R-ROUTE-03 : confond linguistique = L1 seulement
        # --------------------------------------------------------
        if l1_flag and language_confounded(family_row):
            records.append({
                CHILD_ID: child_id,
                TARGET_COLUMN: "Reassess in home language — no orientation yet",
                "Reference_Rule": "R-ROUTE-03",
                "Reference_Item": "L1"
            })
            continue

        # --------------------------------------------------------
        # R-ROUTE-04 : literacy
        # --------------------------------------------------------
        age = get_age(family_row)
        literacy_route = False

        if literacy_flag:
            if pd.isna(age):
                records.append({
                    CHILD_ID: child_id,
                    TARGET_COLUMN: "Review needed — age required for literacy routing",
                    "Reference_Rule": "R-ROUTE-04",
                    "Reference_Item": "R1/R2"
                })
                continue

            if age < 6:
                records.append({
                    CHILD_ID: child_id,
                    TARGET_COLUMN: "Monitor — reassess after age 6",
                    "Reference_Rule": "R-ROUTE-04",
                    "Reference_Item": "R1/R2"
                })
                continue

            literacy_route = True

        # --------------------------------------------------------
        # R-ROUTE-07 : >=2 domaines indépendants
        # --------------------------------------------------------
        independent_domains = sum([
            attention_flag,
            language_flag,
            literacy_route,
            social_flag
        ])

        if independent_domains >= 2:
            records.append({
                CHILD_ID: child_id,
                TARGET_COLUMN: "Multidisciplinary developmental assessment center",
                "Reference_Rule": "R-ROUTE-07",
                "Reference_Item": ""
            })
            continue

        # --------------------------------------------------------
        # R-ROUTE-01 : Attention
        # --------------------------------------------------------
        if attention_flag:
            records.append({
                CHILD_ID: child_id,
                TARGET_COLUMN: "Pediatrician / child psychiatrist",
                "Reference_Rule": "R-ROUTE-01",
                "Reference_Item": "A1/A2"
            })
            continue

        # --------------------------------------------------------
        # R-ROUTE-02 : Language
        # --------------------------------------------------------
        if language_flag:
            records.append({
                CHILD_ID: child_id,
                TARGET_COLUMN: "Speech-language pathologist / ENT",
                "Reference_Rule": "R-ROUTE-02",
                "Reference_Item": "L1/L2"
            })
            continue

        # --------------------------------------------------------
        # R-ROUTE-04 : Literacy
        # --------------------------------------------------------
        if literacy_route:
            records.append({
                CHILD_ID: child_id,
                TARGET_COLUMN: "Neuropsychologist (Dys assessment)",
                "Reference_Rule": "R-ROUTE-04",
                "Reference_Item": "R1/R2"
            })
            continue

        # --------------------------------------------------------
        # R-ROUTE-05 : Social
        # --------------------------------------------------------
        if social_flag:
            records.append({
                CHILD_ID: child_id,
                TARGET_COLUMN: "Developmental pediatrician / TSA team",
                "Reference_Rule": "R-ROUTE-05",
                "Reference_Item": "S1/S2"
            })
            continue

        # --------------------------------------------------------
        # R-ROUTE-06 : Motor isolé
        # --------------------------------------------------------
        if motor_flag:
            records.append({
                CHILD_ID: child_id,
                TARGET_COLUMN: "Occupational therapist",
                "Reference_Rule": "R-ROUTE-06",
                "Reference_Item": "M1/M2"
            })
            continue

        # --------------------------------------------------------
        # DEFAULT
        # --------------------------------------------------------
        records.append({
            CHILD_ID: child_id,
            TARGET_COLUMN: "No orientation — routine monitoring",
            "Reference_Rule": "R-ROUTE-DEFAULT",
            "Reference_Item": ""
        })

    return pd.DataFrame(records)

reference_results = build_reference_result()

print("=" * 72)
print("REFERENCE_RESULT FINAL")
print("=" * 72)
display(reference_results)

print("\nRépartition :")
display(reference_results[TARGET_COLUMN].value_counts(dropna=False))
```

## Construction du dataset ML au niveau enfant

Correction importante : on ne construit plus le dataset ML uniquement à partir de `df_scores` du Sheet 2.  
Les enfants présents seulement dans le Sheet 3 sont conservés, et les scores de domaine sont agrégés à partir des tables professionnelles disponibles.

```python
# ============================================================
# SCORES DE DOMAINES AU NIVEAU ENFANT — TOUTES LES SOURCES
# ============================================================

def compute_domain_scores_for_table(df, table_name):
    cid = first_existing(df, CHILD_ID_CANDIDATES)
    if cid is None:
        return pd.DataFrame()

    out = pd.DataFrame()
    out[CHILD_ID] = df[cid].map(normalize_child_id)

    # Pour chaque domaine, on utilise les questions réellement présentes
    # dans cette table.
    domain_items = {
        "Attention": ["A1", "A2"],
        "Langage": ["L1", "L2"],
        "Lecture_Ecriture": ["R1", "R2"],
        "Social_Comportement": ["S1", "S2"],
        "Motricite": ["M1", "M2"]
    }

    for domain, items in domain_items.items():
        cols = []
        for item in items:
            t, col = resolved_items.get(item, (None, None))
            if t == table_name and col in df.columns:
                cols.append(col)

        if cols:
            numeric = df[cols].apply(pd.to_numeric, errors="coerce")
            out[domain] = numeric.mean(axis=1, skipna=True)
        else:
            out[domain] = np.nan

    return out

score_frames = []

for table_name, df in [
    ("df2_clean", df2_clean),
    ("df3_clean", df3_clean)
]:
    part = compute_domain_scores_for_table(df, table_name)
    if not part.empty:
        score_frames.append(part)

if not score_frames:
    raise ValueError("❌ Aucun score professionnel disponible pour le ML.")

all_scores = pd.concat(score_frames, ignore_index=True)

child_features = (
    all_scores
    .replace("", np.nan)
    .groupby(CHILD_ID)[FEATURE_COLUMNS]
    .mean()
    .reset_index()
)

ml_dataset = child_features.merge(
    reference_results[
        [CHILD_ID, TARGET_COLUMN, "Reference_Rule", "Reference_Item"]
    ],
    on=CHILD_ID,
    how="inner"
)

print("✅ Dataset ML final :", ml_dataset.shape)
display(ml_dataset)

print("\nDistribution des classes détaillées :")
display(ml_dataset[TARGET_COLUMN].value_counts(dropna=False))
```

```python
# ============================================================
# PREPARER X ET y — CLASSE STABLE
# ============================================================

work = ml_dataset[FEATURE_COLUMNS + [TARGET_COLUMN]].copy()

for col in FEATURE_COLUMNS:
    work[col] = pd.to_numeric(work[col], errors="coerce")

work = work.dropna(subset=[TARGET_COLUMN]).copy()

def reference_to_class(value):
    s = normalize_text(value)

    if "needs clarification" in s:
        return "Needs clarification"
    if "monitor" in s and "situational" in s:
        return "Monitor"
    if "reassess in home language" in s:
        return "Language reassessment"
    if "reassess after age 6" in s:
        return "Monitor"
    if "multidisciplinary" in s:
        return "Multidisciplinary assessment"
    if "pediatrician / child psychiatrist" in s:
        return "Attention referral"
    if "speech-language" in s or "ent" in s:
        return "Language referral"
    if "neuropsychologist" in s:
        return "Literacy referral"
    if "developmental pediatrician" in s or "tsa team" in s:
        return "Social referral"
    if "occupational therapist" in s:
        return "Motor referral"
    if "no orientation" in s or "routine monitoring" in s:
        return "No orientation"
    if "review needed" in s:
        return "Review needed"

    return "Other"

work["Reference_Class"] = work[TARGET_COLUMN].map(reference_to_class)

X = work[FEATURE_COLUMNS].copy()
y = work["Reference_Class"].astype(str)

print("X :", X.shape)
print("y :", y.shape)
print("\nClasses ML :")
display(y.value_counts(dropna=False))
```

## Contrôle de validité ML

Deux enfants et deux classes permettent de vérifier le pipeline, mais **pas** d'estimer sérieusement la performance d'un modèle. Le notebook ne fabrique donc ni données ni score de généralisation.

```python
# ============================================================
# CONTRÔLE AVANT ENTRAÎNEMENT
# ============================================================

n_samples = len(X)
n_classes = y.nunique()
class_counts = y.value_counts()

print(f"Nombre d'enfants : {n_samples}")
print(f"Nombre de classes : {n_classes}")
print("Minimum par classe :", int(class_counts.min()) if n_classes else 0)

# Seuil conservateur pour autoriser une vraie évaluation.
MIN_SAMPLES_FOR_ML = 20
MIN_SAMPLES_PER_CLASS = 5

training_ready = (
    n_samples >= MIN_SAMPLES_FOR_ML
    and n_classes >= 2
    and int(class_counts.min()) >= MIN_SAMPLES_PER_CLASS
)

if training_ready:
    print("✅ Dataset suffisamment fourni pour lancer l'évaluation ML.")
else:
    print("⏳ ML non entraîné : volume actuel insuffisant pour une évaluation fiable.")
    print(
        f"   Il faut au minimum {MIN_SAMPLES_FOR_ML} enfants, "
        f"2 classes et {MIN_SAMPLES_PER_CLASS} enfants par classe."
    )

print("training_ready =", training_ready)
```

```python
# ============================================================
# IMPORTS SCIKIT-LEARN — MODÈLES ML
# ============================================================

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

print("✅ Imports ML chargés")
```

```python
# ============================================================
# ENTRAÎNEMENT ML — UNIQUEMENT SI LES DONNÉES LE PERMETTENT
# ============================================================

models = {
    "Logistic Regression": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("model", LogisticRegression(
            max_iter=2000,
            random_state=42,
            class_weight="balanced"
        ))
    ]),
    "Random Forest": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("model", RandomForestClassifier(
            n_estimators=300,
            random_state=42,
            class_weight="balanced"
        ))
    ])
}

results_df = pd.DataFrame()
best_model = None
best_model_name = None
use_test_split = False

if training_ready:
    X_train, X_test, y_train, y_test = train_test_split(
        X, y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )
    use_test_split = True

    results = []

    for name, model in models.items():
        model.fit(X_train, y_train)
        pred = model.predict(X_test)

        results.append({
            "Modèle": name,
            "Évaluation": "TEST",
            "Accuracy (%)": accuracy_score(y_test, pred) * 100,
            "Precision (%)": precision_score(
                y_test, pred, average="macro", zero_division=0
            ) * 100,
            "Recall (%)": recall_score(
                y_test, pred, average="macro", zero_division=0
            ) * 100,
            "F1-score (%)": f1_score(
                y_test, pred, average="macro", zero_division=0
            ) * 100
        })

    results_df = (
        pd.DataFrame(results)
        .sort_values("F1-score (%)", ascending=False)
        .reset_index(drop=True)
    )

    best_model_name = results_df.iloc[0]["Modèle"]
    best_model = models[best_model_name]

    print("=" * 72)
    print("PERFORMANCES ML — JEU DE TEST")
    print("=" * 72)
    display(results_df)

else:
    print("ℹ️ Aucun modèle n'est entraîné avec le petit dataset actuel.")
    print("ℹ️ Le moteur de référence reste la source fonctionnelle de résultat.")
```

```python
# ============================================================
# VALIDATION CROISÉE — UNIQUEMENT SI POSSIBLE
# ============================================================

if training_ready:
    min_class_count = int(y.value_counts().min())
    n_splits = min(5, min_class_count)

    if n_splits >= 2:
        cv = StratifiedKFold(
            n_splits=n_splits,
            shuffle=True,
            random_state=42
        )

        cv_rows = []

        for name, model in models.items():
            scores = cross_val_score(
                model, X, y, cv=cv, scoring="accuracy"
            )

            cv_rows.append({
                "Modèle": name,
                "Accuracy moyenne (%)": scores.mean() * 100,
                "Écart-type (%)": scores.std() * 100
            })

        cv_df = pd.DataFrame(cv_rows)
        display(cv_df)
    else:
        print("⏳ Pas assez d'observations par classe pour la validation croisée.")
else:
    print("⏳ Validation croisée non exécutée.")
```

```python
# ============================================================
# CHEMIN DU MODÈLE ML
# ============================================================

MODEL_PATH = "learning_difficulty_screening_model.joblib"

print("✅ MODEL_PATH =", MODEL_PATH)
```

```python
# ============================================================
# SAUVEGARDE + INFÉRENCE
# ============================================================

if training_ready and best_model is not None:
    artifact = {
        "model": best_model,
        "feature_columns": FEATURE_COLUMNS,
        "target_column": "Reference_Class",
        "reference_result_column": TARGET_COLUMN,
        "reference_rules_version": "R-ROUTE-01_to_R-ROUTE-09",
        "version": "3.0"
    }

    joblib.dump(artifact, MODEL_PATH)
    print("✅ Modèle sauvegardé :", MODEL_PATH)
else:
    print("ℹ️ Aucun modèle ML sauvegardé : données insuffisantes.")

def predict_from_scores(
    attention,
    langage,
    lecture_ecriture,
    social_comportement,
    motricite,
    model_path=MODEL_PATH
):
    """
    Inférence ML uniquement si un modèle a réellement été entraîné.
    """
    if not os.path.exists(model_path):
        raise RuntimeError(
            "Aucun modèle ML entraîné. Le dataset actuel est insuffisant "
            "pour une inférence ML fiable."
        )

    artifact = joblib.load(model_path)
    model = artifact["model"]
    feature_columns = artifact["feature_columns"]

    input_data = pd.DataFrame([{
        "Attention": attention,
        "Langage": langage,
        "Lecture_Ecriture": lecture_ecriture,
        "Social_Comportement": social_comportement,
        "Motricite": motricite
    }])[feature_columns]

    prediction = model.predict(input_data)[0]

    result = {"prediction": prediction}

    if hasattr(model, "predict_proba"):
        probabilities = model.predict_proba(input_data)[0]
        result["probabilities"] = dict(
            zip(model.classes_, probabilities)
        )

    return result

print("✅ Fonction d'inférence prête.")
```
